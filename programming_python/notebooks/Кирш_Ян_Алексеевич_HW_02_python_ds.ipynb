## Задание 1. Класс `Dish` (1,4 баллов)

Реализуйте класс `Dish`, который описывает блюдо в системе доставки еды.

Класс должен быть создан с использованием конструктора `__init__`.

Каждое блюдо должно иметь следующие **атрибуты**:

* `name` (строка) — название блюда;
* `price` (число с плавающей точкой) — цена блюда;
* `is_vegan` (булево значение) — подходит ли блюдо для веганов.

Атрибуты `price` и `is_vegan` должны быть **защищенными** (начинаться с одного подчеркивания), а атрибут `name` — публичным.

Добавьте также в класс:

* метод `get_price()`, возвращающий цену;
* метод `is_suitable_for_vegans()`, возвращающий `True`, если блюдо подходит для веганов, иначе `False`;
* метод `__str__`, возвращающий строку вида:
  `"Название блюда — 250 руб. (веганское)"`
  или
  `"Название блюда — 750 руб. (не веганское)"` (числа никак не округляйте ни в этом, ни в следующих заданиях).

````python
class Dish:
   # НАЧАЛО ВАШЕГО РЕШЕНИЯ
   def __init__(self, name: str, price: float, is_vegan: bool):
      self.name = name
      self._price = price
      self._is_vegan = is_vegan

   def get_price(self) -> float:
      return self._price

   def is_suitable_for_vegans(self) -> bool:
      return self._is_vegan

   def __str__(self) -> str:
      is_vegan_txt = "веганское" if self._is_vegan else "не веганское"
      return f"{self.name} — {self._price} руб. ({is_vegan_txt})"

   # КОНЕЦ ВАШЕГО РЕШЕНИЯ
````

````python
# автотесты, используйте для самопроверки
# (достаточно просто запустить ячейку и убедиться, что
# код срабатывает без ошибок и дойдет до последней строки:))

dish1 = Dish("Салат", 250, True)
assert dish1.name == "Салат"
assert dish1.get_price() == 250
assert dish1.is_suitable_for_vegans() is True
assert str(dish1) == "Салат — 250 руб. (веганское)"

dish2 = Dish("Стейк", 750, False)
assert dish2.get_price() == 750
assert not dish2.is_suitable_for_vegans()
assert str(dish2) == "Стейк — 750 руб. (не веганское)"

print("Все тесты пройдены!")
````

## Задание 2. Наследование и полиморфизм (1,8 баллов)

Создайте два класса: `HotDish` и `ColdDish`, унаследованные от базового класса `Dish`. Эти подклассы должны отражать особенности подачи горячих и холодных блюд.

Каждое блюдо должно при вызове `str()` отображаться в едином стиле:

**Формат строки для всех блюд**:

```
"{название} — {цена} руб. ({тип подачи}, {веганское/не веганское})"
```

Примеры:

* `"Борщ — 150 руб. (горячее, не веганское)"`
* `"Мороженое — 100 руб. (холодное, со льдом, веганское)"`

**Требования:**

1. В классе `HotDish`:

   * добавьте атрибут `temperature` (по умолчанию `"горячее"`),
   * переопределите метод `__str__`, чтобы он включал `temperature`.

2. В классе `ColdDish`:

   * добавьте атрибут `served_with_ice` (булево значение, по умолчанию `False`),
   * переопределите метод `__str__`, чтобы он отображал `"со льдом"` или `"безо льда"` и `"холодное"`.

3. Используйте `super().__str__()` или повторное использование `Dish._is_vegan` для составления строки.

4. Проверьте, что можно создать список из объектов `Dish`, `HotDish`, `ColdDish`, и при итерации по нему корректно работают переопределенные методы `__str__`.


````python
class HotDish(Dish):
    # НАЧАЛО ВАШЕГО РЕШЕНИЯ
    def __init__(self, name: str, price: float, is_vegan: bool, temperature: str = "горячее"):
        super().__init__(name, price, is_vegan)
        self.temperature = temperature
    
    def __str__(self) -> str:
        is_vegan_txt = "веганское" if self._is_vegan else "не веганское"
        return f"{self.name} — {self._price} руб. ({self.temperature}, {is_vegan_txt})"
    # КОНЕЦ ВАШЕГО РЕШЕНИЯ


class ColdDish(Dish):
    # НАЧАЛО ВАШЕГО РЕШЕНИЯ
    def __init__(self, name: str, price: float, is_vegan: bool, served_with_ice: bool = False):
        super().__init__(name, price, is_vegan)
        self.served_with_ice = served_with_ice
    
    def __str__(self) -> str:
        is_vegan_txt = "веганское" if self._is_vegan else "не веганское"
        with_ice_txt = "со льдом" if self.served_with_ice else "безо льда"
        return f"{self.name} — {self._price} руб. (холодное, {with_ice_txt}, {is_vegan_txt})"
    # КОНЕЦ ВАШЕГО РЕШЕНИЯ


# проверка полиморфизма:
dishes = [
    Dish("Хлеб", 10, True),
    HotDish("Суп", 150, False),
    ColdDish("Салат", 120, True, served_with_ice=True)
]

for dish in dishes:
    print(dish)
````

````python
# автотесты

def test_dish_str():
    dish = Dish("Тост", 50, False)
    assert str(dish) == "Тост — 50 руб. (не веганское)"

def test_hotdish_str():
    hot = HotDish("Борщ", 150, False)
    assert str(hot) == "Борщ — 150 руб. (горячее, не веганское)"

def test_colddish_str_with_ice():
    cold = ColdDish("Мороженое", 100, True, served_with_ice=True)
    assert str(cold) == "Мороженое — 100 руб. (холодное, со льдом, веганское)"

def test_colddish_str_without_ice():
    cold = ColdDish("Салат", 120, False)
    assert str(cold) == "Салат — 120 руб. (холодное, безо льда, не веганское)"

def test_polymorphism():
    dishes = [
        Dish("Хлеб", 30, True),
        HotDish("Суп", 120, False),
        ColdDish("Смузи", 150, True, served_with_ice=True)
    ]
    results = [str(d) for d in dishes]
    assert results[0] == "Хлеб — 30 руб. (веганское)"
    assert results[1] == "Суп — 120 руб. (горячее, не веганское)"
    assert results[2] == "Смузи — 150 руб. (холодное, со льдом, веганское)"

test_dish_str()
test_hotdish_str()
test_colddish_str_with_ice()
test_colddish_str_without_ice()
test_polymorphism()
print("Все тесты пройдены!")
````

## Задание 3. Специальные методы `__repr__`, `__eq__` (0,9 баллов)

Дополните ранее реализованный класс `Dish` двумя специальными методами:

* `__repr__(self)` — должен возвращать строку вида:
  `Dish(name='Борщ', price=129.0, is_vegan=False)`

* `__eq__(self, other)` — должен возвращать `True`, если сравниваются два объекта `Dish` с одинаковыми значениями всех атрибутов (`name`, `_price`, `_is_vegan`), иначе `False`.

**Обратите внимание:**
Метод `__eq__` должен проверять, что `other` — также объект класса `Dish`. Если это не так — возвращайте `False`.

````python
class Dish:
    # НАЧАЛО ВАШЕГО РЕШЕНИЯ
    def __init__(self, name: str, price: float, is_vegan: bool):
        self.name = name
        self._price = price
        self._is_vegan = is_vegan

    def get_price(self) -> float:
        return self._price

    def is_suitable_for_vegans(self) -> bool:
        return self._is_vegan

    def __str__(self) -> str:
        is_vegan_txt = "веганское" if self._is_vegan else "не веганское"
        return f"{self.name} — {self._price} руб. ({is_vegan_txt})"

    # дополните класс Dish с первого задания

    def __repr__(self) -> str:
        return f"Dish(name={self.name!r}, price={self._price}, is_vegan={self._is_vegan})"

    def __eq__(self, other: Dish | object) -> bool:
        if not isinstance(other, Dish):
            return False

        return all((
            self.name == other.name,
            self._price == other.get_price(),
            self._is_vegan == other.is_suitable_for_vegans(),
        ))

    # КОНЕЦ ВАШЕГО РЕШЕНИЯ
````

````python
# автотесты

def test_dish_repr():
    d = Dish("Салат", 199.0, True)
    assert repr(d) == "Dish(name='Салат', price=199.0, is_vegan=True)"

def test_dish_eq_same():
    d1 = Dish("Суп", 150.0, False)
    d2 = Dish("Суп", 150.0, False)
    assert d1 == d2

def test_dish_eq_different():
    d1 = Dish("Суп", 150.0, False)
    d2 = Dish("Суп", 150.0, True)
    assert d1 != d2

def test_dish_eq_not_dish():
    d = Dish("Чай", 50.0, True)
    assert d != "Чай"

test_dish_repr()
test_dish_eq_same()
test_dish_eq_different()
test_dish_eq_not_dish()
print("Все тесты пройдены!")
````

## Задание 4. Dataclass для пользователя (1,4 баллов)

Реализуйте класс `Customer`, который представляет пользователя системы доставки еды, с помощью декоратора `@dataclass`.


Класс должен описывать следующую информацию о пользователе:

* `name` (`str`) — имя пользователя;
* `address` (`str`) — адрес доставки;
* `balance` (`float`) — текущий баланс на счете пользователя.

Также добавьте метод `can_pay(amount: float) -> bool`, который  возвращает True, если у клиента достаточно средств, чтобы оплатить `amount`; иначе False.

**Подсказки:**

* Используйте `@dataclass` из модуля `dataclasses`.
* Все поля должны быть публичными.
* Класс должен автоматически поддерживать:

  * методы сравнения (`__eq__`);
  * строковое представление (`__repr__`).
* После создания экземпляра класса должна быть возможность менять значение любого поля напрямую.
* Баланс может быть любым числом с плавающей точкой, включая отрицательные значения.

````python
from dataclasses import dataclass


@dataclass
class Customer:
    # НАЧАЛО ВАШЕГО РЕШЕНИЯ
    name: str
    address: str
    balance: float

    def can_pay(self, amount: float) -> bool:
        return amount <= self.balance

    # КОНЕЦ ВАШЕГО РЕШЕНИЯ
````

````python
# автотесты

def test_customer_repr():
    c = Customer(name="Иван", address="ул. Чехова, д. 5", balance=250.0)
    assert repr(c) == "Customer(name='Иван', address='ул. Чехова, д. 5', balance=250.0)"

def test_customer_fields_access():
    c = Customer(name="Анна", address="ул. Ленина, д. 7", balance=0.0)
    c.balance = 500.0
    c.address = "пр-т Мира, д. 20"
    assert c.balance == 500.0
    assert c.address == "пр-т Мира, д. 20"

def test_customer_equality():
    c1 = Customer(name="Петр", address="ул. Садовая, д. 3", balance=150.0)
    c2 = Customer(name="Петр", address="ул. Садовая, д. 3", balance=150.0)
    c3 = Customer(name="Петр", address="ул. Садовая, д. 3", balance=100.0)
    assert c1 == c2
    assert c1 != c3

def test_customer_can_pay_true():
    c = Customer(name="Bob", address="456 Elm St", balance=50.0)
    assert c.can_pay(49.99) is True
    assert c.can_pay(50.0) is True

def test_customer_can_pay_false():
    c = Customer(name="Charlie", address="789 Oak St", balance=30.0)
    assert c.can_pay(30.01) is False

test_customer_repr()
test_customer_fields_access()
test_customer_equality()
test_customer_can_pay_true()
test_customer_can_pay_false()
print("Все тесты пройдены!")
````

## Задание 5. Класс `Order` и перегрузка операторов (1,8 баллов)

Реализуйте класс `Order`, который представляет заказ клиента в ресторане. Заказ включает список блюд и клиента, который его оформил.

Что нужно по пунктам:

1. Создать класс `Order` с атрибутами:

   * `customer` — объект класса `Customer`.
   * `dishes` — список объектов класса `Dish`, изначально пустой.
2. Реализовать метод `add_dish(self, dish: Dish)`, который добавляет блюдо в заказ.
3. Реализовать метод `total_price(self) -> float`, возвращающий итоговую стоимость всех блюд в заказе.
4. Перегрузить оператор `+`, чтобы можно было складывать два заказа:

   * При сложении создается новый заказ.
   * Клиент в новом заказе — тот же, что и у первого заказа (`self`).
   * Список блюд — объединение блюд из двух заказов.
   * Если объект справа не является заказом, верните `NotImplemented`.
5. Реализовать метод `pay(self)`, который:

   * Проверяет, что заказ не пуст (иначе `AssertionError` с сообщением `"Пустой заказ!"`).
   * Считает итоговую сумму и проверяет, хватает ли у клиента средств через метод `can_pay()` (если нет — выбрасывает `ValueError` с сообщением `"Недостаточно средств!"`).
   * Списывает сумму с баланса клиента и печатает сообщение вида:
     `Иван оплатил 750 руб.`

**Подсказки:**

* Вы можете использовать метод `get_price()` из класса `Dish`.
* Не забудьте про проверку типа в методе `__add__`.

````python
class Order:
    # НАЧАЛО ВАШЕГО РЕШЕНИЯ
    def __init__(self, customer: Customer, dishes: list[Dish] | None = None):
        self.customer = customer
        self.dishes = list(dishes) if dishes is not None else []

    def add_dish(self, dish: Dish) -> None:
        self.dishes.append(dish)

    def total_price(self) -> float:
        total = 0
        for d in self.dishes:
            total += d.get_price()
        return total

    def pay(self):
        assert self.dishes, "Пустой заказ!"

        total = self.total_price()
        if not self.customer.can_pay(total):
            raise ValueError("Недостаточно средств!")

        self.customer.balance -= total
        print(f"{self.customer.name} оплатил {total} руб.")

    def __add__(self, other: object) -> "Order":
        if not isinstance(other, Order):
            return NotImplemented
        return type(self)(self.customer, self.dishes + other.dishes)

    # КОНЕЦ ВАШЕГО РЕШЕНИЯ
````

````python
# автотесты

c = Customer("Иван", "ул. Пушкина", 1000)
d1 = Dish("Пицца", 400, False)
d2 = Dish("Салат", 250, True)
d3 = Dish("Напиток", 100, True)

order1 = Order(c)
order1.add_dish(d1)
order1.add_dish(d2)
assert order1.total_price() == 650

order2 = Order(c)
order2.add_dish(d3)

order3 = order1 + order2
assert isinstance(order3, Order)
assert len(order3.dishes) == 3
assert order3.total_price() == 750

order3.pay()
assert c.balance == 1000 - 750

# проверка на пустой заказ
try:
    empty_order = Order(c)
    empty_order.pay()
    assert False, "Должен быть AssertionError"
except AssertionError as e:
    assert str(e) == "Пустой заказ!"

# проверка на недостаток средств
poor_customer = Customer("Аня", "ул. Ленина", 100)
order_poor = Order(poor_customer)
order_poor.add_dish(Dish("Стейк", 500, False))
try:
    order_poor.pay()
    assert False, "Должен быть ValueError"
except ValueError as e:
    assert str(e) == "Недостаточно средств!"

print("Все тесты пройдены!")
````

## Задание 6. Вызываемый заказ (0,9 баллов)

Сделайте так, чтобы объекты класса `Order` можно было **вызвать как функцию**, например: `order()`.

При вызове объект должен **распечатывать чек** в формате:

```
Чек для [имя клиента]
---------------------
[название блюда 1] - [цена 1] руб.
[название блюда 2] - [цена 2] руб.
...
Итого: [сумма] руб.
```

Если список блюд пуст, выведите сообщение `"Заказ пуст."`.

````python
class Order:
    # НАЧАЛО ВАШЕГО РЕШЕНИЯ
    def __init__(self, customer: Customer, dishes: list[Dish] | None = None):
        self.customer = customer
        self.dishes = list(dishes) if dishes else []

    def add_dish(self, dish: Dish) -> None:
        self.dishes.append(dish)

    def total_price(self) -> float:
        total = 0
        for d in self.dishes:
            total += d.get_price()
        return total

    def pay(self):
        assert self.dishes, "Пустой заказ!"

        total = self.total_price()
        if not self.customer.can_pay(total):
            raise ValueError("Недостаточно средств!")

        self.customer.balance -= total
        print(f"{self.customer.name} оплатил {total} руб.")

    def __add__(self, other: object) -> Order:
        if not isinstance(other, Order):
            return NotImplemented
        return type(self)(self.customer, self.dishes + other.dishes)
    # дополните класс Order с предыдущего задания

    def __call__(self):
        if not self.dishes:
            print("Заказ пуст.")
            return
            
        dishes_txt = "\n".join(f"{d.name} - {d.get_price()} руб." for d in self.dishes)
        print(
            f"Чек для {self.customer.name}\n"
            "---------------------\n"
            f"{dishes_txt}\n"
            f"Итого: {self.total_price()} руб."
        )

    # КОНЕЦ ВАШЕГО РЕШЕНИЯ
````

1. Сделайте так, чтобы по объекту `Order` можно было **итерироваться** в цикле `for`. При этом каждая итерация должна возвращать очередное блюдо из заказа (`Dish`).

    print(dish.name)

   > Подсказка: нужно реализовать специальный метод `__iter__`.

2. Добавьте **свойство** `dish_names`, которое возвращает список **названий** всех блюд в заказе. Сделайте его с помощью декоратора `@property`.

*Пример:*

```python
print(order.dish_names)  # ['Пицца', 'Суп', 'Чай']
```

````python
from collections.abc import Iterator


class Order:
    # НАЧАЛО ВАШЕГО РЕШЕНИЯ
    def __init__(self, customer: Customer, dishes: list[Dish] | None = None):
        self.customer = customer
        self.dishes = list(dishes) if dishes else []

    def add_dish(self, dish: Dish) -> None:
        self.dishes.append(dish)

    def total_price(self) -> float:
        total = 0
        for d in self.dishes:
            total += d.get_price()
        return total

    def pay(self):
        assert self.dishes, "Пустой заказ!"

        total = self.total_price()
        if not self.customer.can_pay(total):
            raise ValueError("Недостаточно средств!")

        self.customer.balance -= total
        print(f"{self.customer.name} оплатил {total} руб.")

    def __add__(self, other: object) -> Order:
        if not isinstance(other, Order):
            return NotImplemented
        return type(self)(self.customer, self.dishes + other.dishes)

    def __call__(self):
        if not self.dishes:
            print("Заказ пуст.")
            return
            
        dishes_txt = "\n".join(f"{d.name} - {d.get_price()} руб." for d in self.dishes)
        print(
            f"Чек для {self.customer.name}\n"
            "---------------------\n"
            f"{dishes_txt}\n"
            f"Итого: {self.total_price()} руб."
        )

    @staticmethod
    def is_discount_day(day: str) -> bool:
        discount_days = ["Monday", "Friday"]
        return day in discount_days

    @classmethod
    def from_dishes(cls, customer: Customer, dishes: list[Dish]) -> 'Order':
        return cls(customer, dishes)

    # дополните класс Order с предыдущего задания
    def __iter__(self) -> Iterator[Dish]:
        return iter(self.dishes)

    @property
    def dish_names(self) -> list[str]:
        return [n.name for n in self.dishes]

    # КОНЕЦ ВАШЕГО РЕШЕНИЯ
````

````python
# автотесты
def test_dish_names_property():
    customer = Customer("Борис", "проспект Гагарина", 500)
    dish1 = Dish("Пицца", 300, False)
    dish2 = Dish("Чай", 50, True)
    order = Order(customer)
    order.add_dish(dish1)
    order.add_dish(dish2)

    assert order.dish_names == ["Пицца", "Чай"]


def test_empty_order_iteration():
    customer = Customer("Виктор", "улица Гоголя", 300)
    order = Order(customer)
    assert list(order) == []


def test_property_on_empty_order():
    customer = Customer("Дина", "площадь Сенная", 200)
    order = Order(customer)
    assert order.dish_names == []

test_dish_names_property()
test_empty_order_iteration()
test_property_on_empty_order()
print("Все тесты пройдены!")
````
