Напишите функцию `swap_places()`, которая принимает в качестве аргумента список lst и меняет местами его первый и последний элементы.

Функция должна возвращать **обновленный** список. Гарантируется, что в исходном списке есть минимум два элемента.

**Примеры работы программы:**
```python
print(swap_places([1, 2, 3]))
## [3, 2, 1]
print(swap_places([1, 2, 3, 4, 5]))
## [5, 2, 3, 4, 1]
print(swap_places(['н', 'л', 'о', 'с']))
## ['с', 'л', 'о', 'н']
```

````python
# НАЧАЛО ВАШЕГО РЕШЕНИЯ
import typing


def swap_places(lst: list[typing.Any]) -> list[typing.Any]:
    lst[-1], lst[0] = lst[0], lst[-1]
    return lst

print(swap_places([1, 2, 3]))
print(swap_places([1, 2, 3, 4, 5]))
print(swap_places(['н', 'л', 'о', 'с']))

# КОНЕЦ ВАШЕГО РЕШЕНИЯ
````

## Задание 5 (1,1 баллов)

Напишите функцию `holes_count()`, которая принимает на вход число number, подсчитывает общее количество отверстий в заданном числе и возвращает результат.

Условимся, что цифрами с отверстиями будем считать 0, 4, 6, 8 и 9. Причем в цифре 8 отверстий два, а в остальных — по одному. Обратите внимание, что функция должна возвращать суммарное количество отверстий по всем цифрам в числе.

**Примеры работы программы:**
```python
print(holes_count(8))
# 2
print(holes_count(146))
# 2
print(holes_count(84628))
# 6
```

````python
# НАЧАЛО ВАШЕГО РЕШЕНИЯ

def holes_count(n: int) -> int:
    n = abs(n)
    if n == 0:
        return 1

    has_holes = {0: 1, 4: 1, 6: 1, 8: 2, 9: 1}
    s = 0
    while n != 0:
        t = n % 10
        if amount := has_holes.get(t):
            s += amount
        n //= 10
    return s

print(holes_count(8))
# 2
print(holes_count(146))
# 2
print(holes_count(84628))
# 6

# КОНЕЦ ВАШЕГО РЕШЕНИЯ
````

Двумерные таблицы (матрицы) — основной формат данных, с которым мы будем работать. Давайте рассмотрим операцию сложения двух матриц. Напишите функцию `matrix_sum()`, которая получает на вход две матрицы и возвращает их сумму.
Примечание. Чтобы сложить две матрицы, нужно просуммировать их соответствующие элементы.
```
1 2 3   2 3 4   1+2 2+3 3+4   3 5 7
2 3 4 + 4 5 6 = 2+4 2+5 4+6 = 6 8 10
5 6 7   4 3 2   5+4 6+3 7+2   9 9 9
```
Важно отметить, что суммировать можно только матрицы одинаковой размерности. Поэтому сначала необходимо проверить, что размеры матриц совпадают, то есть матрицы должны иметь равное количество строк, а в каждой строке должно быть равное количество элементов (столбцов).

Если размеры матриц не совпадают, необходимо вывести на экран сообщение `"Error! Matrices dimensions are different!"` с помощью функции `print()`, а затем вернуть значение `None` из функции `matrix_sum()`.

Гарантируется, что на вход подаются только прямоугольные матрицы (матрицы, в которых количество элементов в каждой из строк одинаково).

**Примеры работы программы:**
```python
matrix_example = [
          [1, 5, 4],
          [4, 2, -2],
          [7, 65, 88]
]

print(matrix_sum(matrix1=matrix_example, matrix2=matrix_example))
# [[2, 10, 8], [8, 4, -4], [14, 130, 176]]

matrix1_example = [
          [1, 5, 4],
          [4, 2, -2]
]

matrix2_example = [
          [10, 15, 43],
          [41, 2, -2],
          [7, 5, 7]
]

print(matrix_sum(matrix1=matrix1_example, matrix2=matrix2_example))
# Error! Matrices dimensions are different!
# None
```

````python
matrix_example = [
          [1, 5, 4],
          [4, 2, -2],
          [7, 65, 88]
]

# НАЧАЛО ВАШЕГО РЕШЕНИЯ

def matrix_sum(matrix1: list[list[int]], matrix2: list[list[int]]) -> list[list[int]] | None:
    if len(matrix1) != len(matrix2) or len(matrix1[0]) != len(matrix2[0]):
        print("Error! Matrices dimensions are different!")
        return None

    res_matrix = []
    for i in range(len(matrix1)):
        row = []
        for j in range(len(matrix1[0])):
            row.append(matrix1[i][j] + matrix2[i][j])
        res_matrix.append(row)
    return res_matrix

matrix_example = [
          [1, 5, 4],
          [4, 2, -2],
          [7, 65, 88]
]

print(matrix_sum(matrix1=matrix_example, matrix2=matrix_example))
# [[2, 10, 8], [8, 4, -4], [14, 130, 176]]

matrix1_example = [
          [1, 5, 4],
          [4, 2, -2]
]

matrix2_example = [
          [10, 15, 43],
          [41, 2, -2],
          [7, 5, 7]
]

print(matrix_sum(matrix1=matrix1_example, matrix2=matrix2_example))
# Error! Matrices dimensions are different!
# None

# КОНЕЦ ВАШЕГО РЕШЕНИЯ
````

* приводит текст к нижнему регистру;
* возвращает список из слов в тексте.

Текст, к которому необходимо применить предобработку, передается в функцию в виде аргумента text.
Для решения задачи вы можете использовать готовый список со знаками препинания:
```python
punctuation_list = ['.', ',', ';', ':', '...', '!', '?', '-', '"', '(', ')']
```
Сделайте этот список локальной переменной функции `get_word_list()`.

**Пример работы программы:**
```python
text_example = "Arrakis, the planet known as Dune, is forever his place."

print(get_words_list(text=text_example))
# ['arrakis', 'the', 'planet', 'known', 'as', 'dune', 'is', 'forev
```

````python
text_example = "Arrakis, the planet known as Dune, is forever his place."

# НАЧАЛО ВАШЕГО РЕШЕНИЯ

def get_words_list(text: str):
    punctuation_list = ['.', ',', ';', ':', '...', '!', '?', '-', '"', '(', ')']
    for p in punctuation_list:
        text = text.replace(p, " ")
    return text.lower().split()

print(get_words_list(text=text_example))
# ['arrakis', 'the', 'planet', 'known', 'as', 'dune', 'is', 'forever', 'his', 'place']

# КОНЕЦ ВАШЕГО РЕШЕНИЯ
````

cheese_data = {
    'чеддер': [370, 5000, 33, 'твердый'],
    'пармезан': [510, 4000, 29, 'твердый'],
    'гауда': [250, 3700, 27, 'полутвердый'],

````python
cheese_data = {
    'чеддер': [370, 5000, 33, 'твердый'],
    'пармезан': [510, 4000, 29, 'твердый'],
    'гауда': [250, 3700, 27, 'полутвердый'],
    'эдам': [220, 10000, 30, 'полутвердый'],
    'горгонзола': [320, 3000, 32, 'полумягкий'],
    'рокфор': [340, 15000, 31, 'полумягкий'],
    'стилтон': [360, 7000, 35, 'полумягкий'],
    'камамбер': [250, 8000, 24, 'мягкий'],
    'бри': [310, 6500, 28, 'мягкий'],
}



# НАЧАЛО ВАШЕГО РЕШЕНИЯ

def filter_by_fat(data: dict[str, list], n: int) -> dict[str, int]:
    res = {}
    for k, v in data.items():
        d = v[-2]
        if d < n:
            res[k] = d
    return res

print(filter_by_fat(cheese_data, 30))
# {'пармезан': 29, 'гауда': 27, 'камамбер': 24, 'бри': 28}
print(filter_by_fat(cheese_data, n=25))
# {'камамбер': 24}
            
# КОНЕЦ ВАШЕГО РЕШЕНИЯ
````

3. *Считайте данные из файла `data.txt`.* Прочитайте содержимое файла построчно.
4. *Проверьте, что каждая строка файла содержит только число.*
5. *Вычислите сумму всех чисел из файла и выведите результат на экран.*

````python
# try/except будет позже
# def is_number(s):
#     """Проверяет, можно ли строку преобразовать в число (int или float)."""
#     try:
#         float(s)
#         return True
#     except ValueError:
#         return False

def is_number(s: str) -> bool:
    """Проверяет, является ли строка числом (целым или десятичным)."""
    if not s:
        return False
    if s[0] in '+-':
        s = s[1:]
    if not s:
        return False
    if s.count('.') > 1:
        return False
    return s.replace('.', '', 1).isdigit()
# НАЧАЛО ВАШЕГО РЕШЕНИЯ


def get_user_input() -> list[str]:
    res = []
    while True:
        n = input()
        if n == "":
            break
        res.append(f"{n}\n")
    return res

def save_user_input(numbers: list[str]):
    with open("data.txt", "w") as f:
        f.writelines(numbers)

def read_file() -> list[float]:
    data = []
    with open("data.txt", "r") as f:
        for line in f:
            d = line.replace("\n", "")
            if is_number(d):
                data.append(float(d))
    return data

user_input = get_user_input()
# print(user_input)
save_user_input(user_input)

data = read_file()
# print(data)

print(sum(data))

# КОНЕЦ ВАШЕГО РЕШЕНИЯ
````
